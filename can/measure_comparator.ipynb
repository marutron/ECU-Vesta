```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import preprocessing
from matplotlib.gridspec import GridSpec
```

```python
# Сравниваем ID в коротких и длинных экспериментах, делаем вывод о возможности коротких измерений по 10 сек
```

```python
long_no_ecu = pd.read_csv("parsed/csv/first_measurement/car_off-efi_off.csv", delimiter="|")
short_no_ecu = pd.read_csv("parsed/csv/10_sec_measurement/no_ecu.csv", delimiter="|")

long_with_ecu = pd.read_csv("parsed/csv/first_measurement/car_off-efi_on.csv", delimiter="|")
short_with_ecu = pd.read_csv("parsed/csv/10_sec_measurement/no_ecu.csv", delimiter="|")

long_ign = pd.read_csv("parsed/csv/first_measurement/car_off-efi_on-ign_on.csv", delimiter="|")
short_ign = pd.read_csv("parsed/csv/10_sec_measurement/ign.csv", delimiter="|")
```

```python
long_no_ecu_id = np.sort(long_no_ecu["ID"].unique())
short_no_ecu_id = np.sort(short_no_ecu["ID"].unique())
no_ecu_diff1 = np.setdiff1d(long_no_ecu_id, short_no_ecu_id) # выводит элементы, которые есть в 1-м массиве, но нет во 2-м
no_ecu_diff2 = np.setdiff1d(short_no_ecu_id, long_no_ecu_id)
print(no_ecu_diff1)
print(no_ecu_diff2)
```

```python
long_with_ecu_id = np.sort(long_with_ecu["ID"].unique())
short_with_ecu_id = np.sort(short_with_ecu["ID"].unique())
with_ecu_diff1 = np.setdiff1d(long_with_ecu_id, short_with_ecu_id)
with_ecu_diff2 = np.setdiff1d(short_with_ecu_id, long_with_ecu_id)
print(with_ecu_diff1)
print(with_ecu_diff2)
```

```python
long_ign_id = np.sort(long_ign["ID"].unique())
short_ign_id = np.sort(short_ign["ID"].unique())
ign_diff1 = np.setdiff1d(long_ign_id, short_ign_id)
ign_diff2 = np.setdiff1d(short_ign_id, long_ign_id)
print(ign_diff1)
print(ign_diff2)
```

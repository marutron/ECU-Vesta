```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import preprocessing
from matplotlib.gridspec import GridSpec
```

```python
no_ecu = pd.read_csv("parsed/csv/10_sec_measurement/no_ecu.csv", delimiter="|")
with_ecu = pd.read_csv("parsed/csv/10_sec_measurement/no_ecu.csv", delimiter="|")
ign = pd.read_csv("parsed/csv/10_sec_measurement/ign.csv", delimiter="|")
work = pd.read_csv("parsed/csv/10_sec_measurement/car_work.csv", delimiter="|")
```

```python
# все id в сети без ECU
all_no_ecu_id = np.sort(no_ecu["ID"].unique())
all_no_ecu_id
```

```python
# все id в сети с подключенным ECU
all_with_ecu_id = np.sort(with_ecu["ID"].unique())
all_with_ecu_id
```

```python
# все id в сети с включенным зажиганием
all_ign_id = np.sort(ign["ID"].unique())
all_ign_id
```

```python
# все id в сети с включенным двигателем
all_work_id = np.sort(work["ID"].unique())
all_work_id
```

```python
# -----------------------------------------------------------------------------
```

```python
# id, производимые ECU без зажигания
ecu_ids = np.setdiff1d(all_with_ecu_id, all_no_ecu_id)
ecu_ids
```

```python
# дополнительные id, производимые при включении зажигания
ign_ids = np.setdiff1d(all_ign_id, all_with_ecu_id)
ign_ids
```

```python
# дополнительные id, передаваемые в сеть при включении двигателя
work_ids = np.setdiff1d(all_work_id, all_ign_id)
work_ids
```

```python
# команды, передаваемые ECU в сеть при включении зажигания
ign_df = ign[ign["ID"].isin(ign_ids)]
ign_commands = ign_df.groupby("ID")["data"].apply(lambda x: x.unique().tolist())

for can_id, group in ign_commands.items():
    print(f"ID: {can_id}")
    print(group)
    print()
```

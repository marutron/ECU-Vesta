```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import preprocessing
from matplotlib.gridspec import GridSpec
```

```python
off_no_ecu = pd.read_csv("parsed/csv/car_off-efi_off-media_off.csv", delimiter="|")
off_with_ecu = pd.read_csv("parsed/csv/car_off-efi_on.csv", delimiter="|")
ign_on = pd.read_csv("parsed/csv/car_off-efi_off-ign_on.csv", delimiter="|")
```

```python
no_ecu_id = np.sort(off_no_ecu["ID"].unique())
no_ecu_id
```

```python
with_ecu_id = np.sort(off_with_ecu["ID"].unique())
with_ecu_id
```

```python
with_ign_id = np.sort(ign_on["ID"].unique())
with_ign_id
```

```python
ecu_ids = np.setdiff1d(with_ecu_id, no_ecu_id)
ecu_ids
```

```python
ign_ids = np.setdiff1d(with_ign_id, with_ecu_id)
ign_ids
```

```python
ecu_df = off_with_ecu[off_with_ecu["ID"].isin(ecu_ids)]
```

```python
ecu_commands = ecu_df.groupby("ID")["data"].apply(lambda x: x.unique().tolist())

for can_id, group in ecu_commands.items():
    print(f"ID: {can_id}")
    print(group)
    print()
```

```python
ign_df = ign_on[ign_on["ID"].isin(ign_ids)]
```

```python
ign_commands = ign_df.groupby("ID")["data"].apply(lambda x: x.unique().tolist())

for can_id, group in ign_commands.items():
    print(f"ID: {can_id}")
    print(group)
    print()
```
